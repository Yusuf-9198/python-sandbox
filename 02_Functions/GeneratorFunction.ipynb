```python
from symtable import Function

from numpy.ma.core import arange


# Generator Function
def fib(n):
    a,b= 0,1
    for i in range(n):
        yield a # it optimized memory utilization
        a,b = b,a+b

fib(7) # <generator object fib at 0x000002396BBBBC40>
for i in fib(10):
    print(i)
```
