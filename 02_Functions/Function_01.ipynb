```python
l = [1,5,6,8,9,"ghazi" , [4,5,76,7,2,8]]
def intReturn(l):
    """This is return integer list""" # Doc String
    l1 = []
    for i in l:
        if type(i) == int or type(i) == float:
            l1.append(i)
    return l1

intReturn(l) # [1, 5, 6, 8, 9]
```

```python
def DynamicPara(*args):
    return args
DynamicPara(4,5) # (4, 5)
DynamicPara("helo",5,6,4,56) # ('helo', 5, 6, 4, 56)
```

```python
def DynamicPara(*yusuf): # args => yusuf
    return yusuf
DynamicPara(4,5) # (4, 5)
DynamicPara("helo",5,6,4,56) # ('helo', 5, 6, 4, 56)
```
