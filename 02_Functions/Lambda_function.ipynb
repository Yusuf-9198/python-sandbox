```python
# power of a number
def power(n,p):
    return n**p
power(4,5)
```

```python
# power by lambda function
a = lambda n,p:n**p
a(4,5)
```

```python
# calsius to f
f = lambda c: (9/5)*c +32
f(27)
```

```python
find_max = lambda x,y: x if x>y else y
find_max(6,8)
```

```python
find_len = lambda s : len(s)
find_len("yusuf")
```
