```python
l = [3,4,5,6,7]
def squre(l):
    l1=[]
    for i in l:
        l1.append(i**2)
    return l1
squre(l)
```

```python
def sq(i):
    return i**2
```

```python
list(map(sq,l)) # [9, 16, 25, 36, 49]
```

```python
list(map(lambda i : i**2,l)) # [9, 16, 25, 36, 49]
```

```python
l1 = [1,2,3,4,5]
l2 = [6,7,8,9,10]
list(map(lambda x,y: x+y,l1,l2)) # [7, 9, 11, 13, 15]
```

```python
s = "yusuf"
list(map(lambda s: s.upper(),s)) # ['Y', 'U', 'S', 'U', 'F']
```

```python
# Reduce
from functools import reduce
```

```python
l = [3,4,5,6,7]
reduce(lambda x,y: x+y,l) # 25
# reduce(lambda x,y,z: x+y+z,l) # error
reduce(lambda x,y:x+y,[1]) # exception case , 1
```

```python
# Max ele in list
reduce(lambda x,y:x if x>y else y,l)
```

```python
# Filter
# find even number in list
list(filter(lambda x : x%2==0,l)) #[4, 6]
```
