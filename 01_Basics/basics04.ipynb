```python
marks = int(input("Enter Marks : "))
# marks = 81
if marks >= 90:
    print("Grade A+")
elif 90 >= marks > 60:
    print("Grade A")
elif 40 <= marks < 60:
    print("grade B")
else :
    print("Failed")
print("Terminated")
```

```python
# Loops
lis = [2,5,7,3,9,68,1,6,788,24,5]
k = 0
for i in lis:
    lis[k] = i+1
    k = k+1
lis # [3, 6, 8, 4, 10, 69, 2, 7, 789, 25, 6]

lis1 = ["yusuf" , "kaif" , "farhan"]
k = 0
for i in lis1:
    lis1[k] = i.upper();
    k = k+1
lis1 # ['YUSUF', 'KAIF', 'FARHAN']
```

```python
lis2 = [1,2,3,4,5,6,"sara",9,5,"zahra"]
l2Str = []
l2Int = []
for i in lis2:
    if(type(i)== int):
        l2Int.append(i)
    else:
        l2Str.append(i)
l2Str # ['sara', 'zahra']
l2Int # [1, 2, 3, 4, 5, 6, 9, 5 ]
```
