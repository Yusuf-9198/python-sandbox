```python
class car:
    def __init__(self,year,make,model,speed):
        self.__year= year # __ => hide this variable from user
        self.__model = model # private variable hai
        self.__make = make
        self.__speed = 0

    def set_speed(self,speed):
        self.__speed = 0 if speed<0 else speed
    def get_speed(self):
        return self.__speed
```

```python
c = car(2021,"BMW","Toyota",21)
# c.year # "can't be excessed"
# # c.__year # "can't be excessed"
c._car__year # 2021
c.set_speed(26)
c.get_speed() # 26
```

```python
class bank_acount:
    def __init__(self,balance):
        self.__balance = balance

    def deposite(self,amount):
        self.__balance = self.__balance + amount

    def withdraw(self,amount):
        if self.__balance >= amount:
            self.__balance = self.__balance - amount
            return True
        else:
            return False

    def get_balance(self):
        return self.__balance
```

```python
ghazi = bank_acount(4500)
```

```python
ghazi.get_balance() # 4500
ghazi.deposite(1000)
ghazi.withdraw(450)
ghazi.get_balance() # 6050
ghazi.withdraw(9000) # False
ghazi.withdraw(50) # True
```
