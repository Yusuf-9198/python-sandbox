```python
class nits:
    def welcome(self):
        # self -> tell that this function bind with this class.
        print("welcome to NIT srinagar")

yusuf = nits()
print(yusuf) # <__main__.nits object at 0x000001FD3565C980>
yusuf.welcome() # welcome to NIT srinagar
```

```python
class college:
    def __init__(self, s_id, s_email,phoneNo): # constructor
        self.phoneNo = phoneNo
        self.s_id = s_id
        self.s_email = s_email
    def return_s_details(self):
        return self.phoneNo,self.s_id,self.s_email
```

```python
yusuf = college(34,"yusf@gmail.com",89564234)
yusuf.return_s_details() # (89564234, 34, 'yusf@gmail.com')
```

```python
class college1:
    def __init__(self, s_id, s_email, phoneNo):
        self.phoneNo1 = phoneNo
        self.s_id1 = s_id
        self.s_email1 = s_email

    def return_s_details(self):
        return self.phoneNo1, self.s_id1, self.s_email1

# if we can change self -> ghazi: it would run
# means "self" is not reserved keyword

kaif = college1(254, "kaif@gmail.com", 89564234)
kaif.return_s_details()  # (89564234, 34, 'yusf@gmail.com')
```

```python
class dataSciece:
    def syllabus(self):
        print("syllaus of data science")
```

```python
class webDev:
    def syllabus(self):
        print("syllaus of web dev")
```

```python
def class_syllabus(class_obj):
    for i in class_obj:
        i.syllabus()
```

```python
DS = dataSciece()
WD = webDev()
class_obj1 = [DS,WD]
class_syllabus(class_obj1)
```
